```python
# Convertendo as colunas "Order Date" e "Ship Date" para o tipo datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
# conferindo os tipos de dados após a conversão
df.dtypes
```

```python
# Padronizando os nomes das colunas do DataFrame
df.columns
```

```python
# Padronizando os nomes das colunas do DataFrame
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)
```

6. Análise Exploratória dos Dados

```python
# Exibir gráfico de boxplot para a coluna "sales" identificar valores discrepantes
sns.boxplot(x=df["sales"])
plt.show()
```

Vendas por Categoria

```python
vendas_categoria = (
    df.groupby("category")["sales"]
      .sum()
      .sort_values(ascending=False)
)

vendas_categoria
```

```python
plt.figure(figsize=(8,5))

plt.bar(vendas_categoria.index, vendas_categoria.values, color=['red', 'yellow', 'orange'])

plt.title("Total de Vendas por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Vendas")

plt.show()
```

Vendas por Segmento

```python
segmentos = (
    df.groupby("segment")["sales"]
      .sum()
      .sort_values(ascending=False)
)

segmentos
```

```python
plt.figure(figsize=(8,5))

plt.bar(segmentos.index, segmentos.values,color='skyblue')

plt.title("Vendas por Segmento")

plt.xlabel("Segmento")
plt.ylabel("Vendas")

plt.show()
```

Relação entre Desconto e Lucro

```python
plt.figure(figsize=(8,6))

plt.scatter(df["discount"], df["profit"], label = 'pontos', marker='*', color='orange', alpha=0.5)

plt.title("Relação entre Desconto e Lucro")

plt.xlabel("Desconto")
plt.ylabel("Lucro")

plt.show()
```
